```python
from auth.auth import google_auth
from scanner.gmail_scanner import scan_gmail
from analyzer.dlp_analyzer import extract_sensitive_data
from alerts import logs
import logging

if __name__ == "__main__":
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s [%(levelname)s] %(message)s",
        handlers=[logging.StreamHandler()]
    )

    services = google_auth()
    gmail = services["gmail"]
    user_email = "main.dupont.tfg@gmail.com" 

    print("Escaneando correos...")
    correos = scan_gmail(gmail)
    print(f"Se han recuperado {len(correos)} correos para análisis.")

    print("Analizando datos sensibles...")
    csv_path = extract_sensitive_data(correos)
    print(f"Análisis completado.")
    print(f"Archivo generado en: {csv_path}")

    logs.run_alerts(gmail, user_email, csv_path, threshold=1)
```
